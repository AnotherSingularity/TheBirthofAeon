```python
import requests

# ⚠️ Replace this with your *current* ngrok tunnel or localhost URL
BASE_URL = "https://08ed-35-190-167-19.ngrok-free.app"

def test_api():
    print("\n🚀 Starting PoC Blockchain Function Tests...\n")

    # ✅ 1. Create Wallets
    print("🔹 Creating Wallets...")
    alice_wallet = requests.post(f"{BASE_URL}/create_wallet", json={"wallet_id": "Alice"}).json()
    bob_wallet = requests.post(f"{BASE_URL}/create_wallet", json={"wallet_id": "Bob"}).json()
    print(f"✅ Create Wallet Alice Response: {alice_wallet}")
    print(f"✅ Create Wallet Bob Response: {bob_wallet}")

    # ✅ 2. Check Balances
    print("\n🔹 Checking Balances...")
    alice_balance = requests.get(f"{BASE_URL}/get_balance?wallet_id=Alice").json()
    bob_balance = requests.get(f"{BASE_URL}/get_balance?wallet_id=Bob").json()
    print(f"✅ Get Balance (Alice) Response: {alice_balance}")
    print(f"✅ Get Balance (Bob) Response: {bob_balance}")

    # ✅ 3. Transfer Tokens
    print("\n🔹 Transferring Tokens...")
    transfer_response = requests.post(
        f"{BASE_URL}/transfer",
        json={
            "sender": "Alice",
            "receiver": "Bob",
            "amount": 100
        }
    ).json()
    print(f"✅ Transfer Response: {transfer_response}")

    # ✅ 4. Charge Gas Fee
    print("\n🔹 Charging Gas Fee...")
    gas_fee_response = requests.post(
        f"{BASE_URL}/charge_gas_fee",
        json={
            "user": "Alice",
            "fee": 0.01
        }
    ).json()
    print(f"✅ Charge Gas Fee Response: {gas_fee_response}")

    # ✅ 5. Check Balances After Transfer
    print("\n🔹 Checking Balances After Transfer...")
    updated_alice_balance = requests.get(f"{BASE_URL}/get_balance?wallet_id=Alice").json()
    updated_bob_balance = requests.get(f"{BASE_URL}/get_balance?wallet_id=Bob").json()
    print(f"✅ Updated Balance (Alice): {updated_alice_balance}")
    print(f"✅ Updated Balance (Bob): {updated_bob_balance}")

    # ✅ 6. Lock BTC for Wrapped BTC (wBTC)
    print("\n🔹 Locking BTC for Wrapped BTC (wBTC)...")
    lock_btc_response = requests.post(
        f"{BASE_URL}/lock_btc",
        json={
            "user": "Alice",
            "amount": 0.5
        }
    ).json()
    print(f"✅ Lock BTC Response: {lock_btc_response}")

    # ✅ 7. Unlock BTC
    print("\n🔹 Unlocking BTC...")
    unlock_btc_response = requests.post(
        f"{BASE_URL}/unlock_btc",
        json={
            "user": "Alice",
            "amount": 0.5
        }
    ).json()
    print(f"✅ Unlock BTC Response: {unlock_btc_response}")

    # ✅ 8. Add a New Block to the Blockchain
    print("\n🔹 Adding a New Block to Blockchain...")
    add_block_response = requests.post(
        f"{BASE_URL}/add_block",
        json={
            "data": "New Test Block",
            "contributor": "Alice",
            "transactions": [],
            "gas_fees": 0.01
        }
    ).json()
    print(f"✅ Add Block Response: {add_block_response}")

    # ✅ 9. Fetch the Blockchain State
    print("\n🔹 Fetching Blockchain State...")
    chain_data = requests.get(f"{BASE_URL}/chain").json()
    print(f"✅ Blockchain Data: {chain_data}")

    print("\n🎉 **All tests completed!** 🎉")


if __name__ == "__main__":
    test_api()
```
