```python
import requests

ngrok_url = "https://5084-35-223-64-72.ngrok-free.app"

response = requests.get(f"{ngrok_url}/chain")

print("Status Code:", response.status_code)  # Check if we get 200 OK
print("Raw Response:", response.text)  # Print raw response before JSON parsing

try:
    json_response = response.json()
    print("Updated Blockchain State:", json_response)
except requests.exceptions.JSONDecodeError:
    print("Error: Response is not valid JSON. Possible empty or failed request.")
```

```python
import requests

ngrok_url = "https://caf7-35-223-64-72.ngrok-free.app"

contract_code = """
def store_value(x):
    global stored_value
    stored_value = x

def get_value():
    return stored_value
"""

deploy_data = {"contract_name": "StorageContract", "contract_code": contract_code}
response = requests.post(f"{ngrok_url}/deploy_contract", json=deploy_data)
print("Deploy Response:", response.json())
```

```python
execute_data = {"contract_name": "StorageContract", "function_name": "store_value", "params": [100]}
response = requests.post(f"{ngrok_url}/execute_contract", json=execute_data)
print("Execute Response:", response.json())
```
