```python
from flask import Flask, request, jsonify
from pyngrok import ngrok
import hashlib
import time
import random

# Initialize Flask app
app = Flask(__name__)

# Start ngrok tunnel manually
public_url = ngrok.connect(5000).public_url
print(f" * ngrok tunnel running at {public_url}")

# Define Blockchain class
class Block:
    def __init__(self, index, previous_hash, timestamp, data, proof):
        self.index = index
        self.previous_hash = previous_hash
        self.timestamp = timestamp
        self.data = data
        self.proof = proof
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        block_string = f"{self.index}{self.previous_hash}{self.timestamp}{self.data}{self.proof}"
        return hashlib.sha256(block_string.encode()).hexdigest()

class Blockchain:
    def __init__(self):
        self.chain = [self.create_genesis_block()]
        self.difficulty = 2  # Starting difficulty for Proof of Cognition (PoC)

    def create_genesis_block(self):
        return Block(0, "0", time.time(), "Genesis Block", 0)

    def get_latest_block(self):
        return self.chain[-1]

    def proof_of_cognition(self, previous_proof):
        """AI-driven Proof of Cognition (PoC) algorithm."""
        new_proof = previous_proof + random.randint(1, 10)
        while not self.is_valid_proof(new_proof):
            new_proof += 1
        return new_proof

    def is_valid_proof(self, proof):
        """Validates if proof meets difficulty criteria."""
        return hashlib.sha256(str(proof).encode()).hexdigest()[:self.difficulty] == "0" * self.difficulty

    def add_block(self, data):
        latest_block = self.get_latest_block()
        proof = self.proof_of_cognition(latest_block.proof)
        new_block = Block(
            index=latest_block.index + 1,
            previous_hash=latest_block.hash,
            timestamp=time.time(),
            data=data,
            proof=proof
        )
        self.chain.append(new_block)
        return new_block

blockchain = Blockchain()

@app.route('/chain', methods=['GET'])
def get_chain():
    chain_data = [{
        "index": block.index,
        "previous_hash": block.previous_hash,
        "timestamp": block.timestamp,
        "data": block.data,
        "proof": block.proof,
        "hash": block.hash
    } for block in blockchain.chain]

    return jsonify({"length": len(chain_data), "chain": chain_data}), 200

@app.route('/new_transaction', methods=['POST'])
def new_transaction():
    """Adds a new transaction to the blockchain using PoC."""
    data = request.get_json()
    required_fields = ["data"]

    if not all(field in data for field in required_fields):
        return jsonify({"error": "Missing fields"}), 400

    new_block = blockchain.add_block(data["data"])

    return jsonify({"message": "New block added", "block": {
        "index": new_block.index,
        "previous_hash": new_block.previous_hash,
        "timestamp": new_block.timestamp,
        "data": new_block.data,
        "proof": new_block.proof,
        "hash": new_block.hash
    }}), 201

# Run Flask app
if __name__ == '__main__':
    app.run(port=5000)

```
