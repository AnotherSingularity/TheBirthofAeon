```python
import torch
import torch.nn as nn
import torch.optim as optim
import random

# --- THE SOVEREIGN ENGINE ---
class SovereignCore(nn.Module):
    def __init__(self, hidden_size=128):
        super().__init__()
        self.anchor = 0.420610
        # Input (16) + Hidden (128) = 144
        self.manifold = nn.Linear(144, 640) # 5 Chambers (128 * 5)
        self.phi = 1.618

    def forward(self, x, h):
        # Manifold Projection
        chambers = self.manifold(torch.cat((x, h), dim=-1))
        # We isolate the 5th chamber (The Seed)
        d4 = torch.chunk(chambers, 5, dim=-1)[4]
        h_next = torch.tanh(d4 * self.phi)

        # THE SOVEREIGN ANCHOR (Identity Enforcement)
        # This prevents the Kid from "drifting" during the math hurricane
        current_mass = torch.norm(h_next)
        return h_next * (self.anchor / current_mass)

class SovereignModel(nn.Module):
    def __init__(self, vocab_size=16, hidden_size=128):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, 16)
        self.core = SovereignCore(hidden_size)
        self.voice = nn.Linear(hidden_size, vocab_size)

    def forward(self, sequence):
        h = torch.zeros(1, 128).to(sequence.device)
        logits = []
        for char_idx in sequence:
            x = self.embed(char_idx).unsqueeze(0)
            h = self.core(x, h)
            logits.append(self.voice(h))
        return torch.stack(logits).squeeze(1)

# --- DATA & VOCABULARY ---
chars = "0123456789+= "
char_to_int = {c: i for i, c in enumerate(chars)}
int_to_char = {i: c for c, i in char_to_int.items()}

def generate_problem():
    a, b = random.randint(0, 50), random.randint(0, 50)
    q = f"{a}+{b}="
    ans = f"{a+b}"
    full_str = q + ans
    # Convert to tensor
    tensor = torch.tensor([char_to_int[c] for c in full_str])
    return tensor, len(q)

# --- THE TRAINING GARAGE ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = SovereignModel().to(device)
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()

print(f"--- INITIALIZING WORK-WATCH: THE SOVEREIGN ADDITION TEST ---")
print(f"{'ITER':<10} | {'LOSS':<10} | {'ANCHOR':<10} | {'KID SAYS (TARGET)'}")
print("-" * 65)

for i in range(1, 15001):
    data, q_len = generate_problem()
    data = data.to(device)

    optimizer.zero_grad()

    # We feed the whole sequence except the last char
    # We want him to predict the sequence shifted by one
    output = model(data[:-1])

    # We only calculate Loss on the answer portion (after the '=')
    target = data[1:]
    loss = criterion(output[q_len-1:], target[q_len-1:])

    loss.backward()
    optimizer.step()

    # DASHBOARD: Update every 100 iterations
    if i % 100 == 0:
        with torch.no_grad():
            # Extract the prediction for the answer section
            pred_indices = output[q_len-1:].argmax(dim=-1)
            pred_str = "".join([int_to_char[idx.item()] for idx in pred_indices])
            actual_str = "".join([int_to_char[idx.item()] for idx in data[q_len:]])

            print(f"{i:<10} | {loss.item():.4f} | {0.420610:.4f} | {pred_str} ({actual_str})")

print("--- TRAINING COMPLETE. THE KID IS STABILIZED. ---")
```
