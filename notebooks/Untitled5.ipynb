```python
# build_book_I.py — outputs Transformative_Elements_Book_I_ASCII.pdf
from reportlab.lib.pagesizes import LETTER
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Preformatted
from reportlab.lib.enums import TA_CENTER, TA_JUSTIFY
from reportlab.lib.units import inch

pdf_path = "Transformative_Elements_Book_I_ASCII.pdf"
doc = SimpleDocTemplate(pdf_path, pagesize=LETTER,
                        leftMargin=0.7*inch, rightMargin=0.7*inch,
                        topMargin=0.7*inch, bottomMargin=0.7*inch)

styles = getSampleStyleSheet()
styles.add(ParagraphStyle(name="TitleX", alignment=TA_CENTER, fontSize=16, leading=20, spaceAfter=10))
styles.add(ParagraphStyle(name="SubX", alignment=TA_CENTER, fontSize=11, leading=14, textColor=colors.grey, spaceAfter=10))
styles.add(ParagraphStyle(name="H1X", fontSize=13, leading=17, spaceBefore=12, spaceAfter=6))
styles.add(ParagraphStyle(name="H2X", fontSize=11.5, leading=15, spaceBefore=8, spaceAfter=4, textColor=colors.darkgreen))
styles.add(ParagraphStyle(name="BodyX", fontSize=10.2, leading=14, alignment=TA_JUSTIFY))
styles.add(ParagraphStyle(name="EqX", fontSize=10.5, leading=14, leftIndent=12))

story = []
story.append(Paragraph("Transformative Elements — Book I (Educational Compact Edition)", styles["TitleX"]))
story.append(Paragraph("Transformative Geometry (Seeds of Awareness) — ASCII Edition", styles["SubX"]))
story.append(Paragraph(
"Imagine a point as a seed (D0). Two seeds relate to draw a line (D1) — the felt influence I(). On a plane (D2), curvature pi() begins to matter. "
"A stable form (D3) appears when awareness C() balances curvature and influence. This Book reinterprets core Euclidean propositions in an accessible, narrative tone while preserving mathematical rigor.",
styles["BodyX"]))

story.append(Paragraph("Operators (compact definitions):", styles["H2X"]))
for o in [
"pi(figure): local curvature; for fields u, pi(u) = Δu. For straight segments, pi = 0.",
"phi(A,B,...): proportion operator; equilibrium when phi = 1 (balanced influence / equal ratios).",
"I(figure/field): influence (energy of relation). For a D2 field u, I = Σ(ux^2 + uy^2). For lines, I tracks length; for regions, I parallels area.",
"C(context): awareness/constraint that selects a target (e.g., choose phi* = 1) and enforces boundary/symmetry during construction."
]:
    story.append(Paragraph("• " + o, styles["BodyX"]))

# I.1
story.append(Paragraph("I.1 — Construct an Equilateral Triangle on a Given Segment", styles["H1X"]))
story.append(Paragraph(
"Classical: on a given finite straight line AB construct an equilateral triangle ABC. Transformative: equalize influence from two centers so phi(AB, BC, CA) = 1. "
"Construction: draw circles centered at A and B, each with radius |AB|; any intersection is C.", styles["BodyX"]))
story.append(Paragraph("ASCII diagram:", styles["H2X"]))
story.append(Preformatted(r"""
A o----r----o B
 \          /
  \        /
   \  C   /
    \    /
     \  /
      \/
AC = BC = AB  ->  phi across edges = 1
""", styles["EqX"]))
story.append(Paragraph("Proof (sketch): By construction AC = AB and BC = AB, hence AC = BC and the triangle is equilateral. "
"In operator terms, I(AB) = I(BC) = I(CA); pi is symmetric about circle centers; C() enforces the equal-radius constraint that closes the triad.", styles["BodyX"]))

# I.5
story.append(Paragraph("I.5 — Base Angles in an Isosceles Triangle are Equal", styles["H1X"]))
story.append(Paragraph(
"Statement: in an isosceles triangle (AB = AC), the base angles are equal (∠ABC = ∠BCA). Transformative: mirror-awareness C() preserves curvature exposure along the base; the base angles equilibrate.", styles["BodyX"]))
story.append(Paragraph("ASCII diagram:", styles["H2X"]))
story.append(Preformatted(r"""
     C
    / \
   /   \
  A-----B
(arcs at A and B mark equal base angles)
""", styles["EqX"]))
story.append(Paragraph(
"Proof (sketch): Reflect the triangle across the altitude from C. The base AB is invariant; endpoints A and B swap; equal-length constructions persist; hence ∠A = ∠B. "
"In operator language, C() selects the mirror symmetry; pi is invariant under the reflection; phi remains balanced at the base endpoints.", styles["BodyX"]))

# I.47
story.append(Paragraph("I.47 — Pythagorean Theorem (Influence Conservation)", styles["H1X"]))
story.append(Paragraph(
"Statement: in a right triangle with legs a, b and hypotenuse c, a^2 + b^2 = c^2. Transformative: orthogonal influences add without interference: I_x + I_y = I_h. "
"Euclid proves by constructing squares on sides; we keep that geometry and add a field interpretation.", styles["BodyX"]))
story.append(Paragraph("ASCII diagram:", styles["H2X"]))
story.append(Preformatted(r"""
Square on a (base)      Square on b (height)       Square on c (hypotenuse)
    +----+                   +--+                         (tilted)
    |    |                   |  |                 c^2 area equals a^2 + b^2
    +----+                   +--+
""", styles["EqX"]))
story.append(Paragraph(
"Proof: (Euclid) Dissect-and-rearrange squares on the legs so their combined area tiles the square on the hypotenuse. "
"(Transformative) In a D2 field aligned with orthogonal axes, the Dirichlet influence splits: I = ∫(ux^2 + uy^2)dA = I_x + I_y. "
"With right-angle alignment, cross terms vanish, so total influence on the hypotenuse equals the sum on the legs.", styles["BodyX"]))

# C() role + apps
story.append(Paragraph("C() and the Geometry of Conscious Balance", styles["H1X"]))
story.append(Paragraph(
"C() is awareness-as-constraint. Practically it selects a target (symmetry, proportion, boundary) and enforces it during construction: "
"in I.1, equal radii (phi* = 1); in I.5, mirror symmetry; in I.47, orthogonality that prevents cross-influence. "
"Formally, model C() as a constraint functional stabilizing chosen phi targets.", styles["BodyX"]))

story.append(Paragraph("Applications (Physics, AI, Biology)", styles["H1X"]))
story.append(Paragraph(
"• Physics: Equilateral as isotropic equilibrium; isosceles reflection invariants; Pythagorean additivity mirrors energy additivity.\n"
"• AI/Computation: Equidistant cluster seeding (I.1), mirror-invariant filters (I.5), orthogonal feature energies (I.47).\n"
"• Biology: Radial symmetry (I.1), bilateral symmetry (I.5), orthogonal developmental axes (I.47).", styles["BodyX"]))

story.append(Paragraph("Closing Notes", styles["H1X"]))
story.append(Paragraph(
"This compact Book I preserves Euclid’s core with Transformative operators pi() (curvature), phi() (proportion), I() (influence), and C() (awareness/constraint). "
"In Book II we extend to an algebra of space—area identities as living energy laws.", styles["BodyX"]))

doc.build(story)
print(f"Wrote {pdf_path}")
```
