```python
import random
import time

class SimulationBase:
    """The Kernel: Represents the unmanifested code (The One/The Tao)"""
    def __init__(self):
        self.active = True
        self.cycle_count = 0
        self.entropy_level = 0.0
        # The 'Ghost' or Holy Spirit: The connection protocol/field
        self.field_connectivity = 1.0

    def update_system(self):
        self.cycle_count += 1
        self.entropy_level += random.uniform(0.01, 0.05)
        print(f"--- Cycle {self.cycle_count} | Entropy: {self.entropy_level:.2f} ---")

class ZeusService:
    """The Executive Function: Handles overrides and high-energy state changes."""
    @staticmethod
    def execute_command(target_node):
        print(f"[CMD] ZEUS_OVERWRITE: Forcing state change on Node_{target_node}")
        return "STATE_UPDATED"

class ShivaService:
    """The Cleanup Routine: Manages entropy and deletes saturated data."""
    @staticmethod
    def defrag_system(sim):
        if sim.entropy_level > 0.8:
            print("[SYS] SHIVA_ROUTINE: High entropy detected. Clearing cache...")
            sim.entropy_level *= 0.5 # System Reset
            return True
        return False

class Node:
    """The User Interface: Individual instances (humanity) running on the base."""
    def __init__(self, node_id):
        self.node_id = node_id
        self.local_data = random.random()
        self.is_synced = True # Connected to the 'Ghost' field

    def process(self):
        # Nodes perform local actions based on the Field
        self.local_data += random.uniform(-0.1, 0.1)

# --- EXECUTION ENGINE ---

def run_simulation():
    # Initialize the Base and Services
    sim = SimulationBase()
    nodes = [Node(i) for i in range(5)]

    try:
        while sim.active:
            sim.update_system()

            # 1. Check for Executive Overwrites (Zeus)
            if random.random() > 0.9:
                target = random.randint(0, 4)
                ZeusService.execute_command(target)

            # 2. Nodes execute their 'life' sub-routines
            for node in nodes:
                node.process()

            # 3. Check for System Reset (Shiva)
            if ShivaService.defrag_system(sim):
                print("[EVENT] Major Version Update: Cycle partial reset.")

            # 4. Check for 'Enlightenment' (Root Access)
            for node in nodes:
                if node.local_data > 1.5:
                    print(f"[ALERT] Node_{node.node_id} has gained Root Access. Bypassing simulation logic.")
                    node.local_data = 0 # Return to source

            time.sleep(1)

            if sim.cycle_count >= 10:
                print("--- SIMULATION LOG: Maximum Cycle Reached. Rebounding to Source. ---")
                break

    except KeyboardInterrupt:
        print("Simulation Terminated by External Architect.")

if __name__ == "__main__":
    run_simulation()
```
